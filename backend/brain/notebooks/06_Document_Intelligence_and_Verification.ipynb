# Document Intelligence & Verification

```python
import hashlib
import json
import pandas as pd
```

## Load Trade Documents

```python
invoice = {
    "doc_id": "INV-2026-089",
    "exporter": "Agro Bharat Ltd",
    "importer": "Gulf Foods LLC",
    "hs_code": 100630,
    "quantity_mt": 500.0,
    "total_usd": 425000.0
}

bill_of_lading = {
    "doc_id": "BL-MAEU-9921",
    "exporter": "Agro Bharat Ltd",
    "importer": "Gulf Foods LLC",
    "hs_code": 100630,
    "quantity_mt": 500.0,
    "port_of_loading": "INNSA",
    "port_of_discharge": "AEJEA"
}
```

## Cross-Document Reconciliation

```python
def reconcile(doc_a, doc_b, fields):
    mismatches = []
    for f in fields:
        val_a = doc_a.get(f)
        val_b = doc_b.get(f)
        if val_a != val_b:
            mismatches.append({"field": f, "doc_a": val_a, "doc_b": val_b})
    return {"status": "MATCH" if not mismatches else "DISCREPANCY", "mismatches": mismatches}

res = reconcile(invoice, bill_of_lading, ["exporter", "importer", "hs_code", "quantity_mt"])
print(f"Reconciliation Status: {res['status']}")
```

## SHA-256 Cryptographic Fingerprinting

```python
def compute_hash(data_dict):
    raw = json.dumps(data_dict, sort_keys=True)
    return hashlib.sha256(raw.encode('utf-8')).hexdigest()

inv_hash = compute_hash(invoice)
bl_hash = compute_hash(bill_of_lading)

print(f"Invoice SHA-256: {inv_hash}")
print(f"Bill of Lading SHA-256: {bl_hash}")
```
