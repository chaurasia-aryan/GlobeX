# Destination Country Ranking Engine

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Load Destination Feature Store

```python
data_paths = [
    'data/processed/destination_country_ranking_features.csv',
    '../data/processed/destination_country_ranking_features.csv',
    'backend/brain/data/processed/destination_country_ranking_features.csv',
    'processed/destination_country_ranking_features.csv'
]
path = next(p for p in data_paths if os.path.exists(p))
df = pd.read_csv(path)
df.head()
```

## Normalization & Weighted Scoring

```python
def min_max(s):
    return (s - s.min()) / (s.max() - s.min() + 1e-6)

num_cols = df.select_dtypes(include=[np.number]).columns
scaled_df = df[num_cols].apply(min_max)

df['opportunity_score'] = (scaled_df.mean(axis=1) * 100).round(1)
df['rank'] = df['opportunity_score'].rank(ascending=False).astype(int)
```

## Top Destination Country Leaderboard

```python
country_col = 'destination_country' if 'destination_country' in df.columns else ('partner_iso3' if 'partner_iso3' in df.columns else df.columns[0])
leaderboard = df.sort_values('opportunity_score', ascending=False)[[country_col, 'opportunity_score', 'rank']].head(10)
leaderboard
```
